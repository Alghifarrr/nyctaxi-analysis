```python
query = """
    SELECT
        COUNT(*) AS total_extreme_distance
    FROM yellow_taxi
    WHERE trip_distance > 100
"""

pd.read_sql(query, engine)
```

```python
query = """
    SELECT
        COUNT(*) AS total_extreme_distance,
        ROUND(AVG(trip_distance), 2) AS avg_distance,
        ROUND(MIN(trip_distance), 2) AS min_distance,
        ROUND(MAX(trip_distance), 2) AS max_distance
    FROM yellow_taxi
    WHERE trip_distance > 100
"""

pd.read_sql(query, engine)
```

```python
query = """
    SELECT
        COUNT(*) AS suspicious_trips
    FROM yellow_taxi
    WHERE trip_distance > 100
        AND (tpep_dropoff_datetime - tpep_pickup_datetime) < INTERVAL '3 hours'
"""

pd.read_sql(query, engine)
```

```python
query = """
    SELECT
        tpep_pickup_datetime,
        tpep_dropoff_datetime,
        trip_distance,
        fare_amount,
        total_amount
    FROM yellow_taxi
    WHERE trip_distance > 100
        AND (tpep_dropoff_datetime - tpep_pickup_datetime) >= INTERVAL '3 hours'
    ORDER BY trip_distance DESC
"""

long_trips = pd.read_sql(query, engine)

long_trips
```

```python
query = """
    SELECT
        EXTRACT(HOUR FROM tpep_pickup_datetime) AS pickup_hour,
        COUNT(*) AS total_trips
    FROM yellow_taxi
    GROUP BY pickup_hour
    ORDER BY pickup_hour
"""

hourly_trips = pd.read_sql(query, engine)

hourly_trips
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    hourly_trips["pickup_hour"],
    hourly_trips["total_trips"]
)

plt.title("Number of Trips by Pickup Hour")
plt.xlabel("Pickup Hour")
plt.ylabel("Number of Trips")

plt.xticks(range(24))
plt.show()
```

```python
query = """
    SELECT
        EXTRACT(DOW FROM tpep_pickup_datetime) AS day_of_week,
        COUNT(*) AS total_trips
    FROM yellow_taxi
    GROUP BY day_of_week
    ORDER BY day_of_week
"""

daily_trips = pd.read_sql(query, engine)

daily_trips
```

```python
day_names = {
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday"
}

daily_trips["day_name"] = daily_trips["day_of_week"].map(day_names)

daily_trips
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    daily_trips["day_name"],
    daily_trips["total_trips"]
)

plt.title("Number of Trips by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Number of Trips")

plt.show()
```

```python
query = """
    SELECT
        CASE
            WHEN EXTRACT(DOW FROM tpep_pickup_datetime) IN (0, 6)
                THEN 'Weekend'
            ELSE 'Weekday'
        END AS day_type,
        COUNT(*) AS total_trips
    FROM yellow_taxi
    GROUP BY day_type
    ORDER BY day_type
"""

weekday_weekend = pd.read_sql(query, engine)

weekday_weekend
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    weekday_weekend["day_type"],
    weekday_weekend["total_trips"]
)

plt.title("Number of Trips: Weekday vs Weekend")
plt.xlabel("Day Type")
plt.ylabel("Number of Trips")

plt.show()
```

```python
query = """
    SELECT
        trip_distance,
        total_amount
    FROM yellow_taxi
    WHERE trip_distance > 0
        AND total_amount > 0
    ORDER BY RANDOM()
    LIMIT 100000
"""

distance_revenue = pd.read_sql(query, engine)

distance_revenue.head()
```

```python
plt.figure(figsize=(10, 6))

plot_data = distance_revenue[
    (distance_revenue["trip_distance"] <= 20) &
    (distance_revenue["total_amount"] <= 100)
]

plt.scatter(
    plot_data["trip_distance"],
    plot_data["total_amount"],
    alpha=0.3
)

plt.title("Trip Distance vs Total Amount (≤20 miles, ≤$100)")
plt.xlabel("Trip Distance (miles)")
plt.ylabel("Total Amount ($)")

plt.show()
```

```python
query = """
    SELECT
        EXTRACT(HOUR FROM tpep_pickup_datetime) AS pickup_hour,
        COUNT(*) AS total_trips,
        SUM(total_amount) AS total_revenue,
        AVG(total_amount) AS average_trip_value
    FROM yellow_taxi
    WHERE total_amount > 0
    GROUP BY pickup_hour
    ORDER BY pickup_hour
"""

hourly_revenue = pd.read_sql(query, engine)

hourly_revenue
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    hourly_revenue["pickup_hour"],
    hourly_revenue["total_revenue"]
)

plt.title("Total Revenue by Pickup Hour")
plt.xlabel("Pickup Hour")
plt.ylabel("Total Revenue ($)")

plt.xticks(range(24))
plt.show()
```

```python
query = """
    SELECT
        EXTRACT(DOW FROM tpep_pickup_datetime) AS day_of_week,
        COUNT(*) AS total_trips,
        SUM(total_amount) AS total_revenue,
        AVG(total_amount) AS average_trip_value
    FROM yellow_taxi
    WHERE total_amount > 0
    GROUP BY day_of_week
    ORDER BY day_of_week
"""

daily_revenue = pd.read_sql(query, engine)

daily_revenue
```

```python
day_names = {
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday"
}

daily_revenue["day_name"] = daily_revenue["day_of_week"].map(day_names)

daily_revenue
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    daily_revenue["day_name"],
    daily_revenue["total_revenue"]
)

plt.title("Total Revenue by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Total Revenue ($)")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    daily_revenue["day_name"],
    daily_revenue["average_trip_value"]
)

plt.title("Average Trip Value by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Average Trip Value ($)")

plt.show()
```

```python
query = """
    SELECT
        EXTRACT(EPOCH FROM (
            tpep_dropoff_datetime - tpep_pickup_datetime
        )) / 60 AS trip_duration_minutes,
        trip_distance,
        total_amount
    FROM yellow_taxi
    WHERE trip_distance > 0
        AND total_amount > 0
    ORDER BY RANDOM()
    LIMIT 100000
"""

duration_data = pd.read_sql(query, engine)

duration_data.head()
```

```python
duration_data["trip_duration_minutes"].describe()
```

```python
plt.figure(figsize=(10, 5))

plot_duration = duration_data[
    duration_data["trip_duration_minutes"] <= 120
]

plt.hist(
    plot_duration["trip_duration_minutes"],
    bins=50
)

plt.title("Distribution of Trip Duration (≤120 minutes)")
plt.xlabel("Trip Duration (minutes)")
plt.ylabel("Number of Trips")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plot_duration = duration_data[
    (duration_data["trip_duration_minutes"] <= 120) &
    (duration_data["total_amount"] <= 100)
]

plt.scatter(
    plot_duration["trip_duration_minutes"],
    plot_duration["total_amount"],
    alpha=0.3
)

plt.title("Trip Duration vs Total Amount")
plt.xlabel("Trip Duration (minutes)")
plt.ylabel("Total Amount ($)")

plt.show()
```

```python
query = """
    SELECT
        tpep_pickup_datetime,
        tpep_dropoff_datetime,
        trip_distance,
        total_amount,
        EXTRACT(EPOCH FROM (
            tpep_dropoff_datetime - tpep_pickup_datetime
        )) / 60 AS trip_duration_minutes
    FROM yellow_taxi
    WHERE trip_distance > 100
        OR total_amount > 500
    ORDER BY trip_distance DESC
    LIMIT 100
"""

anomaly_data = pd.read_sql(query, engine)

anomaly_data.head(10)
```
